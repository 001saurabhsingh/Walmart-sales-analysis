# NumPy Operations - Walmart Sales Analysis
Practicing array operations, aggregations, percentiles, and outlier detection with NumPy.

```python
import numpy as np
import pandas as pd
```

### 1. Load Data into NumPy Arrays

```python
df = pd.read_csv('data/Walmart_Sales.csv')
sales = df['Weekly_Sales'].to_numpy()

print("Array shape:", sales.shape)
print("Data type:", sales.dtype)
print("First 5 elements:", sales[:5])
```

### 2. Basic Statistical Summary

```python
print(f"Total Sales:    ${np.sum(sales):,.2f}")
print(f"Mean Sales:     ${np.mean(sales):,.2f}")
print(f"Median Sales:   ${np.median(sales):,.2f}")
print(f"Std Deviation:  ${np.std(sales):,.2f}")
print(f"Min Sales:      ${np.min(sales):,.2f}")
print(f"Max Sales:      ${np.max(sales):,.2f}")
```

### 3. Percentiles & Interquartile Range (IQR)

```python
q25 = np.percentile(sales, 25)
q50 = np.percentile(sales, 50)
q75 = np.percentile(sales, 75)
iqr = q75 - q25

print(f"25th Percentile (Q1): ${q25:,.2f}")
print(f"50th Percentile (Q2): ${q50:,.2f}")
print(f"75th Percentile (Q3): ${q75:,.2f}")
print(f"IQR:                 ${iqr:,.2f}")
```

### 4. Boolean Masking & Outlier Detection

```python
lower_fence = q25 - 1.5 * iqr
upper_fence = q75 + 1.5 * iqr

outlier_mask = (sales < lower_fence) | (sales > upper_fence)
outlier_sales = sales[outlier_mask]

print(f"Lower Fence:        ${lower_fence:,.2f}")
print(f"Upper Fence:         ${upper_fence:,.2f}")
print(f"Number of Outliers: {len(outlier_sales)}")
print(f"Peak Outlier Value:  ${np.max(outlier_sales):,.2f}")
```

### 5. Array Normalization (Min-Max Scaling)

```python
norm_sales = (sales - np.min(sales)) / (np.max(sales) - np.min(sales))

print("Normalized min:", np.min(norm_sales))
print("Normalized max:", np.max(norm_sales))
print("Sample normalized values:", np.round(norm_sales[:5], 4))
```
