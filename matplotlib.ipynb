# Matplotlib & Seaborn Visualizations - Walmart Sales
Plotting line graphs, bar charts, distributions, boxplots, and heatmaps.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set plot aesthetics
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
```

### 1. Load Data & Format Dates

```python
df = pd.read_csv('data/Walmart_Sales.csv')
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df['Month'] = df['Date'].dt.month
df.head()
```

### 2. Line Chart: Weekly Sales Trend (2010 - 2012)

```python
weekly_totals = df.groupby('Date', as_index=False)['Weekly_Sales'].sum()

plt.figure(figsize=(12, 5))
plt.plot(weekly_totals['Date'], weekly_totals['Weekly_Sales'], color='darkgreen', linewidth=1.8)
plt.title('Total Weekly Sales Trend Across All Stores')
plt.xlabel('Date')
plt.ylabel('Total Sales ($)')
plt.tight_layout()
plt.show()
```

### 3. Bar Chart: Average Sales by Month

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly_avg = df.groupby('Month')['Weekly_Sales'].mean().reset_index()
monthly_avg['Month_Name'] = months

plt.figure(figsize=(10, 5))
sns.barplot(data=monthly_avg, x='Month_Name', y='Weekly_Sales', color='steelblue')
plt.title('Average Weekly Sales by Month')
plt.xlabel('Month')
plt.ylabel('Average Weekly Sales ($)')
plt.tight_layout()
plt.show()
```

### 4. Horizontal Bar Chart: Top 10 Stores

```python
top_stores = df.groupby('Store')['Weekly_Sales'].mean().nlargest(10).reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(data=top_stores, x='Weekly_Sales', y='Store', orient='h', color='#2b5c8f')
plt.title('Top 10 Stores by Average Weekly Sales')
plt.xlabel('Average Weekly Sales ($)')
plt.ylabel('Store Number')
plt.tight_layout()
plt.show()
```

### 5. Histogram with KDE: Sales Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Weekly_Sales'], bins=35, kde=True, color='teal')
plt.title('Distribution of Weekly Sales')
plt.xlabel('Weekly Sales ($)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

### 6. Boxplot: Detecting Outlier Weeks

```python
plt.figure(figsize=(10, 3.5))
sns.boxplot(x=df['Weekly_Sales'], color='orange')
plt.title('Box Plot of Weekly Sales')
plt.xlabel('Weekly Sales ($)')
plt.tight_layout()
plt.show()
```

### 7. Correlation Heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Numeric Features Correlation Heatmap')
plt.tight_layout()
plt.show()
```
