# Walmart Sales Analysis (2010 - 2012)
Exploratory data analysis of weekly sales across 45 Walmart stores to identify seasonal trends, evaluate store performance, and examine the impact of holiday weeks and external economic factors.

## 1. Setup & Environment
Importing libraries (Pandas, NumPy, Matplotlib, Seaborn) and configuring visualization styles.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

DATA_PATH = Path("data/Walmart_Sales.csv")
OUTPUT_DIR = Path("visualizations")
OUTPUT_DIR.mkdir(exist_ok=True)
```

## 2. Load Dataset
Loading the Walmart sales CSV data from the data directory.

```python
df = pd.read_csv(DATA_PATH)
display(df.head())
```

## 3. Dataset Overview & Summary Statistics
Examining column types, dataset dimensions, and baseline statistical summaries.

```python
print("Shape:", df.shape)
df.info()
display(df.describe().round(2))
```

## 4. Data Quality & Checks
Checking for null values and duplicate records.

```python
display(df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
display(df.dtypes)
```

## 5. Date Parsing & Feature Engineering
Converting the Date column from string (DD-MM-YYYY) to datetime, and extracting Year, Month, and Week number for time-series analysis.

```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y", errors="coerce")
print("Invalid dates:", df["Date"].isna().sum())

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Week"] = df["Date"].dt.isocalendar().week.astype("int64")

display(df.head())
print("Missing values after date conversion:")
display(df.isna().sum())
```

## 6. Overall Sales Summary
Baseline sales metrics across the entire dataset.

```python
total_sales = df["Weekly_Sales"].sum()
average_weekly_sales = df["Weekly_Sales"].mean()
median_weekly_sales = df["Weekly_Sales"].median()

print(f"Total recorded store-week sales: ${total_sales:,.2f}")
print(f"Average store-week sales: ${average_weekly_sales:,.2f}")
print(f"Median store-week sales: ${median_weekly_sales:,.2f}")
print("Stores:", df["Store"].nunique())
print("Date range:", df["Date"].min().date(), "to", df["Date"].max().date())
```

## 7. Store Performance Comparison
Comparing average weekly sales across all 45 stores to identify top and bottom performers.

```python
store_sales = (
    df.groupby("Store")["Weekly_Sales"]
      .agg(Total_Sales="sum", Average_Weekly_Sales="mean", Weeks_Recorded="count")
      .sort_values("Average_Weekly_Sales", ascending=False)
)
display(store_sales.head(10).round(2))

top_stores = store_sales.head(10).reset_index()
sns.barplot(data=top_stores, x="Average_Weekly_Sales", y="Store", color="steelblue")
plt.title("Top 10 Stores by Average Weekly Sales")
plt.xlabel("Average weekly sales ($)")
plt.ylabel("Store")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "01_top_stores.png", dpi=150)
plt.show()
```

## 8. Weekly Sales Trend (2010 - 2012)
Tracking combined weekly sales across stores to observe seasonal patterns and end-of-year holiday spikes.

## 8. Sales trend over time

**Question:** How does combined sales across all stores change from week to week?

We sum sales for each date. A line chart is appropriate because dates have a natural order and we want to see movement over time.

```python
weekly_trend = df.groupby("Date", as_index=False)["Weekly_Sales"].sum().sort_values("Date")
sns.lineplot(data=weekly_trend, x="Date", y="Weekly_Sales", color="darkgreen")
plt.title("Total Weekly Sales Across All Stores")
plt.xlabel("Date")
plt.ylabel("Combined weekly sales ($)")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "02_weekly_sales_trend.png", dpi=150)
plt.show()
```

**How to read the chart:** Peaks indicate weeks when the combined sales of the 45 stores were higher; dips indicate lower-sales weeks. Seasonal peaks may be worth investigating, but a single peak should not automatically be treated as a repeatable trend.

## 10. Sales Distribution
Examining weekly sales spread using histogram and boxplot visualizations.

```python
monthly_sales = df.groupby("Month")["Weekly_Sales"].mean().reindex(range(1, 13))
month_names = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
monthly_plot = monthly_sales.rename("Average_Weekly_Sales").reset_index()
monthly_plot["Month_Name"] = month_names

sns.barplot(data=monthly_plot, x="Month_Name", y="Average_Weekly_Sales", color="cornflowerblue")
plt.title("Average Weekly Sales by Month")
plt.xlabel("Month")
plt.ylabel("Average weekly sales ($)")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "03_monthly_sales.png", dpi=150)
plt.show()
display(monthly_sales.sort_values(ascending=False).round(2))
```

### Outlier Detection (IQR Method)
Calculating 25th percentile, 75th percentile, and the 1.5x IQR upper threshold to isolate high-volume holiday sales weeks.

```python
sns.histplot(df["Weekly_Sales"], bins=35, kde=True, color="teal")
plt.title("Distribution of Weekly Sales")
plt.xlabel("Weekly sales ($)")
plt.ylabel("Number of store-weeks")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "04_sales_distribution.png", dpi=150)
plt.show()

sns.boxplot(x=df["Weekly_Sales"], color="orange")
plt.title("Spread and Potential Outliers in Weekly Sales")
plt.xlabel("Weekly sales ($)")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "05_sales_boxplot.png", dpi=150)
plt.show()
```

## 11. Holiday vs Non-Holiday Analysis
Comparing sales performance during holiday weeks (Super Bowl, Labor Day, Thanksgiving, Christmas) against regular weeks.

```python
sales = df["Weekly_Sales"].to_numpy()
q1 = np.percentile(sales, 25)
q3 = np.percentile(sales, 75)
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

potential_outliers = df[(df["Weekly_Sales"] < lower_fence) | (df["Weekly_Sales"] > upper_fence)]
print(f"Q1: ${q1:,.2f}")
print(f"Q3: ${q3:,.2f}")
print(f"IQR: ${iqr:,.2f}")
print("Potential outlier rows:", len(potential_outliers))
```

## 12. Macroeconomic Correlations
Evaluating the correlation between weekly sales and external variables: Temperature, Fuel Price, CPI, and Unemployment.

```python
holiday_comparison = df.groupby("Holiday_Flag")["Weekly_Sales"].agg(
    Average_Sales="mean", Median_Sales="median", Store_Weeks="count"
)
holiday_comparison.index = ["Non-holiday weeks", "Holiday weeks"]
display(holiday_comparison.round(2))

holiday_plot = holiday_comparison.reset_index(names="Week_Type")
sns.barplot(data=holiday_plot, x="Week_Type", y="Average_Sales", color="mediumpurple")
plt.title("Average Weekly Sales: Holiday vs Non-holiday")
plt.xlabel("")
plt.ylabel("Average weekly sales ($)")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "06_holiday_comparison.png", dpi=150)
plt.show()
```

## 13. Summary Statistics & Key Metrics

```python
numeric_corr = df.select_dtypes(include="number").corr()
sns.heatmap(numeric_corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Between Numeric Variables")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "07_correlation_heatmap.png", dpi=150)
plt.show()

factor_columns = ["Temperature", "Fuel_Price", "CPI", "Unemployment"]
factor_correlations = df[factor_columns + ["Weekly_Sales"]].corr()["Weekly_Sales"].drop("Weekly_Sales")
display(factor_correlations.sort_values(ascending=False).to_frame("Correlation_with_Weekly_Sales"))
```

## 14. Key Findings & Takeaways

### Key Observations:
- **Top & Bottom Stores:** Store 20 records the highest average weekly sales (~.11M), while Store 33 has the lowest (~). Store revenue varies substantially by location.
- **Holiday Lift:** Holiday weeks generate an average sales lift of **+7.8%** compared to non-holiday weeks, with the largest volume spikes occurring around Thanksgiving (Black Friday) and the pre-Christmas week.
- **Monthly Seasonality:** December generates the highest average weekly sales (~.28M), followed by November.
- **Outliers:** 34 store-week entries exceed the upper IQR threshold (.72M), which corresponds to Q4 holiday demand surges rather than data entry anomalies.
- **Economic Indicators:** Fuel price (+0.01), CPI (-0.07), temperature (-0.06), and unemployment (-0.11) show minimal linear correlation with weekly sales in this aggregated dataset.

```python
best_store = store_sales["Average_Weekly_Sales"].idxmax()
best_store_avg = store_sales.loc[best_store, "Average_Weekly_Sales"]
lowest_store = store_sales["Average_Weekly_Sales"].idxmin()
lowest_store_avg = store_sales.loc[lowest_store, "Average_Weekly_Sales"]
best_month = monthly_sales.idxmax()
best_month_avg = monthly_sales.max()
holiday_avg = holiday_comparison.loc["Holiday weeks", "Average_Sales"]
nonholiday_avg = holiday_comparison.loc["Non-holiday weeks", "Average_Sales"]
holiday_pct = (holiday_avg / nonholiday_avg - 1) * 100

print(f"Highest average-sales store: Store {best_store} (${best_store_avg:,.2f} per week)")
print(f"Lowest average-sales store: Store {lowest_store} (${lowest_store_avg:,.2f} per week)")
print(f"Highest average-sales month: {month_names[best_month - 1]} (${best_month_avg:,.2f})")
print(f"Holiday-week average is {holiday_pct:.1f}% higher/lower than non-holiday average.")
```

Use the results above to complete the project story. Potential business actions should be treated as hypotheses to investigate, not guaranteed improvements.

### Recommendations to consider
1. Review practices at high-average-sales stores and check whether they can be adapted elsewhere.

### Dataset limitations
- The file contains weekly sales and selected context variables, but no profit, cost, inventory, promotion, customer or store-size information.
- The date range is February 2010 to October 2012, so the first and last years are partial.
- There are 143 weekly observations per store in this file; not every calendar week is necessarily represented.
- Holiday_Flag is a broad indicator and does not identify the specific holiday.
- Correlation and group averages are descriptive and do not prove causation.
- Monthly averages combine multiple years and may be affected by uneven coverage and store-level differences.
